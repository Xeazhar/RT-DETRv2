```python
import json
import os

# --- CONFIG ---
base_path = 'C:/Users/jazzb/rt-deterv2/RT-DETRv2/dataset/RDD-Object-Detection-Dataset'
# Process both Train and Valid
splits = ['train', 'valid'] 

mapping = {
    'Potholes': 'Pothole', 'D40': 'Pothole',
    'D20': 'Alligator Crack',
    'D00': 'Crack', 'D01': 'Crack', 'D10': 'Crack', 'D11': 'Crack'
}

new_categories = [
    {"id": 1, "name": "Pothole", "supercategory": "none"},
    {"id": 2, "name": "Alligator Crack", "supercategory": "none"},
    {"id": 3, "name": "Crack", "supercategory": "none"}
]

name_to_new_id = {cat['name']: cat['id'] for cat in new_categories}

for split in splits:
    # 1. Path Setup - NOTICE THE DIFFERENT OUTPUT NAME
    input_file = os.path.join(base_path, split, '_annotations.coco.json')
    output_file = os.path.join(base_path, split, 'refined_annotations.json')
    
    if not os.path.exists(input_file):
        print(f"⚠️ Skipping {split}: File not found at {input_file}")
        continue

    with open(input_file, 'r') as f:
        coco = json.load(f)

    # 2. Map Old IDs to New IDs
    old_id_to_new_id = {}
    for old_cat in coco['categories']:
        if old_cat['name'] in mapping:
            new_name = mapping[old_cat['name']]
            old_id_to_new_id[old_cat['id']] = name_to_new_id[new_name]

    # 3. Filter Annotations
    new_annotations = []
    valid_image_ids = set()
    for ann in coco['annotations']:
        if ann['category_id'] in old_id_to_new_id:
            ann['category_id'] = old_id_to_new_id[ann['category_id']]
            new_annotations.append(ann)
            valid_image_ids.add(ann['image_id'])

    # 4. Filter Images
    new_images = [img for img in coco['images'] if img['id'] in valid_image_ids]

    # 5. Save as a NEW file
    refined_coco = {
        "images": new_images,
        "annotations": new_annotations,
        "categories": new_categories
    }

    with open(output_file, 'w') as f:
        json.dump(refined_coco, f)

    print(f"✅ {split.upper()} saved to: {output_file}")
    print(f"   📊 {len(new_images)} images kept.")
```

```python
import json
import os

# Define your paths based on your previous output
paths = {
    'train': {
        'orig': 'C:/Users/jazzb/rt-deterv2/RT-DETRv2/dataset/RDD-Object-Detection-Dataset/train/refined_annotations.json',
        'sub': 'C:/Users/jazzb/rt-deterv2/RT-DETRv2/dataset/RDD-Object-Detection-Dataset/train/subset_100.json'
    },
    'valid': {
        'orig': 'C:/Users/jazzb/rt-deterv2/RT-DETRv2/dataset/RDD-Object-Detection-Dataset/valid/refined_annotations.json',
        'sub': 'C:/Users/jazzb/rt-deterv2/RT-DETRv2/dataset/RDD-Object-Detection-Dataset/valid/subset_100.json'
    }
}

def create_small_subset(source, destination, limit=100):
    if not os.path.exists(source):
        print(f"❌ Could not find {source}")
        return

    with open(source, 'r') as f:
        data = json.load(f)

    # Slice images and match annotations
    subset_images = data['images'][:limit]
    img_ids = set(img['id'] for img in subset_images)
    subset_anns = [ann for ann in data['annotations'] if ann['image_id'] in img_ids]

    subset_data = {
        'info': data.get('info', {}),
        'licenses': data.get('licenses', []),
        'images': subset_images,
        'annotations': subset_anns,
        'categories': data['categories']
    }

    with open(destination, 'w') as f:
        json.dump(subset_data, f)
    print(f"✅ Subset created: {destination} ({len(subset_images)} images)")

# Run for both train and valid
create_small_subset(paths['train']['orig'], paths['train']['sub'])
create_small_subset(paths['valid']['orig'], paths['valid']['sub'])
```

```python
import torch
import torch.nn as nn
from PIL import Image, ImageDraw
import torchvision.transforms as T
import numpy as np

# 1. Setup paths - CHANGE THESE
MODEL_PATH = 'output/rtdetrv2_r50vd_m_6x_coco/last.pth'
IMAGE_PATH = 'C:/Users/jazzb/rt-deterv2/RT-DETRv2/dataset/RDD-Object-Detection-Dataset/valid/Japan_012810_jpg.rf.77edb20cff4f8bb56db4d0d9e71f7968.jpg' 
OUTPUT_NAME = 'detection_result.jpg'

# 2. Load the model manually (Avoids the 'str' issue)
from src.core import YAMLConfig
import torch

cfg = YAMLConfig('configs/rtdetrv2/rtdetrv2_r50vd_m_7x_coco.yml', resume=MODEL_PATH)
model = cfg.model # Use the raw model architecture
checkpoint = torch.load(MODEL_PATH, map_location='cpu')

# If 'ema' exists in the checkpoint, use it (it's usually more accurate)
if 'ema' in checkpoint:
    state_dict = checkpoint['ema']['module']
elif 'model' in checkpoint:
    state_dict = checkpoint['model']
else:
    state_dict = checkpoint

model.load_state_dict(state_dict)
model.eval()

# 3. Prepare Image
img_pil = Image.open(IMAGE_PATH).convert('RGB')
w, h = img_pil.size
# RT-DETR usually expects 640x640
transforms = T.Compose([
    T.Resize((640, 640)),
    T.ToTensor(),
])
img_tensor = transforms(img_pil).unsqueeze(0) # Add batch dimension

# 4. Run Inference
with torch.no_grad():
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    model.to(device)
    img_tensor = img_tensor.to(device)
    
    outputs = model(img_tensor)
    
    # RT-DETR raw output is often a dict with 'pred_logits' and 'pred_boxes'
    if isinstance(outputs, dict):
        logits = outputs['pred_logits']
        boxes = outputs['pred_boxes']
        # Convert logits to probabilities and get labels
        probs = logits.sigmoid()
        scores, labels = probs.max(-1)
    else:
        # If it's a list/tuple, filter out strings to find the tensors
        tensors = [t for t in outputs if torch.is_tensor(t)]
        # Usually: [labels_tensor, boxes_tensor, scores_tensor]
        if len(tensors) == 3:
            labels, boxes, scores = tensors
        elif len(tensors) == 2:
            labels, boxes = tensors
            scores = torch.ones_like(labels).float()
        else:
            raise ValueError(f"Could not find tensors in output. Types: {[type(o) for o in outputs]}")

# 5. Drawing (Safe Unpacking)
draw = ImageDraw.Draw(img_pil)

# Ensure they are tensors before calling .cpu()
def to_numpy(x):
    if torch.is_tensor(x):
        return x[0].cpu().numpy() # Take first batch
    return x

labels_np = to_numpy(labels)
boxes_np = to_numpy(boxes)
scores_np = to_numpy(scores)

# Road Damage Classes
class_map = {
    0: "Pothole", 
    1: "Alligator Crack", 
    2: "Crack"
}
count = 0

for i in range(len(boxes_np)):
    scr = scores_np[i]
    if scr > 0.1: # confidence score
        count += 1
        # Handle if label is numeric or string
        label_val = labels_np[i]
        label_name = class_map.get(int(label_val), f"CLS_{label_val}")
        
        # [cx, cy, w, h] normalized -> [L, T, R, B]
        cx, cy, bw, bh = boxes_np[i]
        left = (cx - bw/2) * w
        top = (cy - bh/2) * h
        right = (cx + bw/2) * w
        bottom = (cy + bh/2) * h
        
        draw.rectangle([left, top, right, bottom], outline="red", width=4)
        draw.text((left, top - 15), f"{label_name} {scr:.2f}", fill="red")

img_pil.save(OUTPUT_NAME)
print(f"✅ Detection complete! Found {count} objects. Saved to {OUTPUT_NAME}")
```
