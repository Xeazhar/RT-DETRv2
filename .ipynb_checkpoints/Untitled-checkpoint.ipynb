```python
import json
import os

# Paths
input_ann = 'C:/Users/jazzb/rt-deterv2/RT-DETRv2/dataset/RDD-Object-Detection-Dataset/train/_annotations.coco.json'
output_ann = 'C:/Users/jazzb/rt-deterv2/RT-DETRv2/dataset/RDD-Object-Detection-Dataset/train/_annotations.coco.json'

with open(input_ann, 'r') as f:
    coco = json.load(f)

# 1. Define our new category mapping
# Key: Old Name, Value: New Name
mapping = {
    'Potholes': 'Pothole',
    'D40': 'Pothole',
    'D20': 'Alligator Crack',
    'D00': 'Crack',
    'D01': 'Crack',
    'D10': 'Crack',
    'D11': 'Crack'
}

# 2. Create the new Categories list for COCO
new_categories = [
    {"id": 1, "name": "Pothole", "supercategory": "none"},
    {"id": 2, "name": "Alligator Crack", "supercategory": "none"},
    {"id": 3, "name": "Crack", "supercategory": "none"}
]

# Create a lookup for New Name -> New ID
name_to_new_id = {cat['name']: cat['id'] for cat in new_categories}

# Create a lookup for Old ID -> New ID
old_cat_id_to_new_id = {}
for old_cat in coco['categories']:
    old_name = old_cat['name']
    if old_name in mapping:
        new_name = mapping[old_name]
        old_cat_id_to_new_id[old_cat['id']] = name_to_new_id[new_name]

# 3. Filter and Update Annotations
new_annotations = []
valid_image_ids = set()

for ann in coco['annotations']:
    if ann['category_id'] in old_cat_id_to_new_id:
        # Update the category ID to our new 1, 2, or 3
        ann['category_id'] = old_cat_id_to_new_id[ann['category_id']]
        new_annotations.append(ann)
        valid_image_ids.add(ann['image_id'])

# 4. Only keep images that actually have the labels we want
new_images = [img for img in coco['images'] if img['id'] in valid_image_ids]

# 5. Save the refined dataset
refined_coco = {
    "images": new_images,
    "annotations": new_annotations,
    "categories": new_categories
}

with open(output_ann, 'w') as f:
    json.dump(refined_coco, f)

print(f"✅ Refined dataset saved to: {output_ann}")
print(f"📊 Summary: {len(new_images)} images and {len(new_annotations)} annotations kept.")
```

```python
import json
import os

# Define the paths clearly
base_path = 'C:/Users/jazzb/rt-deterv2/RT-DETRv2/dataset/RDD-Object-Detection-Dataset'
splits = ['train', 'valid']
mapping = {'Potholes': 'Pothole', 'D40': 'Pothole', 'D20': 'Alligator Crack', 
           'D00': 'Crack', 'D01': 'Crack', 'D10': 'Crack', 'D11': 'Crack'}
new_categories = [
    {"id": 1, "name": "Pothole", "supercategory": "none"},
    {"id": 2, "name": "Alligator Crack", "supercategory": "none"},
    {"id": 3, "name": "Crack", "supercategory": "none"}
]

for split in splits:
    input_path = os.path.join(base_path, split, '_annotations.coco.json')
    output_path = os.path.join(base_path, split, 'refined_annotations.coco.json')
    
    if not os.path.exists(input_path):
        print(f"❌ Error: Cannot find {input_path}")
        continue

    with open(input_path, 'r') as f:
        coco = json.load(f)

    # Map IDs
    old_to_new = {}
    for cat in coco['categories']:
        if cat['name'] in mapping:
            new_name = mapping[cat['name']]
            old_to_new[cat['id']] = next(c['id'] for c in new_categories if c['name'] == new_name)

    # Filter
    new_anns = []
    valid_img_ids = set()
    for ann in coco['annotations']:
        if ann['category_id'] in old_to_new:
            ann['category_id'] = old_to_new[ann['category_id']]
            new_anns.append(ann)
            valid_img_ids.add(ann['image_id'])

    new_imgs = [img for img in coco['images'] if img['id'] in valid_img_ids]

    with open(output_path, 'w') as f:
        json.dump({"images": new_imgs, "annotations": new_anns, "categories": new_categories}, f)
    
    print(f"✅ Created: {output_path} ({len(new_imgs)} images)")
```
