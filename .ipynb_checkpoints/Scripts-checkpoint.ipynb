```python
import json
import os

# --- CONFIG ---
base_path = 'C:/Users/jazzb/rt-deterv2/RT-DETRv2/dataset/RDD-Object-Detection-Dataset'
# Process both Train and Valid
splits = ['train', 'valid'] 

mapping = {
    'Potholes': 'Pothole', 'D40': 'Pothole',
    'D20': 'Alligator Crack',
    'D00': 'Crack', 'D01': 'Crack', 'D10': 'Crack', 'D11': 'Crack'
}

new_categories = [
    {"id": 1, "name": "Pothole", "supercategory": "none"},
    {"id": 2, "name": "Alligator Crack", "supercategory": "none"},
    {"id": 3, "name": "Crack", "supercategory": "none"}
]

name_to_new_id = {cat['name']: cat['id'] for cat in new_categories}

for split in splits:
    # 1. Path Setup - NOTICE THE DIFFERENT OUTPUT NAME
    input_file = os.path.join(base_path, split, '_annotations.coco.json')
    output_file = os.path.join(base_path, split, 'refined_annotations.json')
    
    if not os.path.exists(input_file):
        print(f"Skipping {split}: File not found at {input_file}")
        continue

    with open(input_file, 'r') as f:
        coco = json.load(f)

    # 2. Map Old IDs to New IDs
    old_id_to_new_id = {}
    for old_cat in coco['categories']:
        if old_cat['name'] in mapping:
            new_name = mapping[old_cat['name']]
            old_id_to_new_id[old_cat['id']] = name_to_new_id[new_name]

    # 3. Filter Annotations
    new_annotations = []
    valid_image_ids = set()
    for ann in coco['annotations']:
        if ann['category_id'] in old_id_to_new_id:
            ann['category_id'] = old_id_to_new_id[ann['category_id']]
            new_annotations.append(ann)
            valid_image_ids.add(ann['image_id'])

    # 4. Filter Images
    new_images = [img for img in coco['images'] if img['id'] in valid_image_ids]

    # 5. Save as a NEW file
    refined_coco = {
        "images": new_images,
        "annotations": new_annotations,
        "categories": new_categories
    }

    with open(output_file, 'w') as f:
        json.dump(refined_coco, f)

    print(f" {split.upper()} saved to: {output_file}")
    print(f"    {len(new_images)} images kept.")
```

```python
import torch
import torch.nn as nn
from PIL import Image, ImageDraw
import torchvision.transforms as T
import numpy as np

# 1. Setup paths - CHANGE THESE
MODEL_PATH = 'output/rtdetrv2_r50vd_m_6x_coco/last.pth'
IMAGE_PATH = 'C:/Users/jazzb/rt-deterv2/RT-DETRv2/dataset/DPWH-sample-Dataset/testing/d0adef8d-4_-__Potholes_16.png' 
OUTPUT_NAME = 'detection_result.jpg'

# 2. Load the model manually (Avoids the 'str' issue)
from src.core import YAMLConfig
import torch

cfg = YAMLConfig('configs/rtdetrv2/rtdetrv2_r50vd_m_7x_coco.yml', resume=MODEL_PATH)
model = cfg.model # Use the raw model architecture
checkpoint = torch.load(MODEL_PATH, map_location='cpu')

# If 'ema' exists in the checkpoint, use it (it's usually more accurate)
if 'ema' in checkpoint:
    state_dict = checkpoint['ema']['module']
elif 'model' in checkpoint:
    state_dict = checkpoint['model']
else:
    state_dict = checkpoint

model.load_state_dict(state_dict)
model.eval()

# 3. Prepare Image
img_pil = Image.open(IMAGE_PATH).convert('RGB')
w, h = img_pil.size
# RT-DETR usually expects 640x640
transforms = T.Compose([
    T.Resize((640, 640)),
    T.ToTensor(),
])
img_tensor = transforms(img_pil).unsqueeze(0) # Add batch dimension

# 4. Run Inference
with torch.no_grad():
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    model.to(device)
    img_tensor = img_tensor.to(device)
    
    outputs = model(img_tensor)
    
    # RT-DETR raw output is often a dict with 'pred_logits' and 'pred_boxes'
    if isinstance(outputs, dict):
        logits = outputs['pred_logits']
        boxes = outputs['pred_boxes']
        # Convert logits to probabilities and get labels
        probs = logits.sigmoid()
        scores, labels = probs.max(-1)
    else:
        # If it's a list/tuple, filter out strings to find the tensors
        tensors = [t for t in outputs if torch.is_tensor(t)]
        # Usually: [labels_tensor, boxes_tensor, scores_tensor]
        if len(tensors) == 3:
            labels, boxes, scores = tensors
        elif len(tensors) == 2:
            labels, boxes = tensors
            scores = torch.ones_like(labels).float()
        else:
            raise ValueError(f"Could not find tensors in output. Types: {[type(o) for o in outputs]}")

# 5. Drawing (Safe Unpacking)
draw = ImageDraw.Draw(img_pil)

# Ensure they are tensors before calling .cpu()
def to_numpy(x):
    if torch.is_tensor(x):
        return x[0].cpu().numpy() # Take first batch
    return x

labels_np = to_numpy(labels)
boxes_np = to_numpy(boxes)
scores_np = to_numpy(scores)

# Road Damage Classes
class_map = {
    0: "Pothole", 
    1: "Alligator Crack", 
    2: "Crack"
}
count = 0

for i in range(len(boxes_np)):
    scr = scores_np[i]
    if scr > 0.1: # confidence score
        count += 1
        # Handle if label is numeric or string
        label_val = labels_np[i]
        label_name = class_map.get(int(label_val), f"CLS_{label_val}")
        
        # [cx, cy, w, h] normalized -> [L, T, R, B]
        cx, cy, bw, bh = boxes_np[i]
        left = (cx - bw/2) * w
        top = (cy - bh/2) * h
        right = (cx + bw/2) * w
        bottom = (cy + bh/2) * h
        
        draw.rectangle([left, top, right, bottom], outline="red", width=4)
        draw.text((left, top - 15), f"{label_name} {scr:.2f}", fill="red")

img_pil.save(OUTPUT_NAME)
print(f" Detection complete! Found {count} objects. Saved to {OUTPUT_NAME}")
```

```python
import json
import os
from collections import Counter

# Path to your DPWH training file
train_json_path = r'C:\Users\jazzb\rt-deterv2\RT-DETRv2\dataset\DPWH-sample-Dataset\train\instances_train.json'

with open(train_json_path, 'r') as f:
    data = json.load(f)

# 1. Check Categories
print("--- Categories Found ---")
categories = {cat['id']: cat['name'] for cat in data['categories']}
for cat_id, cat_name in categories.items():
    print(f"ID: {cat_id} | Name: {cat_name}")

# 2. Count Images
print(f"\nTotal Images: {len(data['images'])}")

# 3. Count Annotations per Class
print("\n--- Distribution of Labels ---")
label_counts = Counter([ann['category_id'] for ann in data['annotations']])

for cat_id, count in label_counts.items():
    name = categories.get(cat_id, "Unknown")
    print(f"{name} (ID {cat_id}): {count} instances")

# 4. Check for Empty Images
annotated_img_ids = {ann['image_id'] for ann in data['annotations']}
all_img_ids = {img['id'] for img in data['images']}
empty_images = all_img_ids - annotated_img_ids
print(f"\nImages without any labels: {len(empty_images)}")
```

```python
import cv2
import json
import os
import matplotlib.pyplot as plt

# --- CONFIGURATION ---
json_path = r'C:\Users\jazzb\rt-deterv2\RT-DETRv2\dataset\DPWH-sample-Dataset\train\instances_train_fixed.json'
img_folder = r'C:\Users\jazzb\rt-deterv2\RT-DETRv2\dataset\DPWH-sample-Dataset\train'
target_image = "62771d83-4_-__Potholes_14.png"

# 1. Load the JSON data
with open(json_path, 'r') as f:
    coco_data = json.load(f)

# 2. Find the Image ID for the target filename
img_id = None
for img in coco_data['images']:
    if img['file_name'] == target_image:
        img_id = img['id']
        break

if img_id is None:
    print(f"❌ Error: Could not find {target_image} in the JSON file.")
    exit()

# 3. Load the image from the folder
full_img_path = os.path.join(img_folder, target_image)
image = cv2.imread(full_img_path)

if image is None:
    print(f"❌ Error: Could not load image from {full_img_path}. Check the path.")
    exit()

# 4. Get all annotations for this image ID
found_boxes = 0
for ann in coco_data['annotations']:
    if ann['image_id'] == img_id:
        # COCO format is [x_min, y_min, width, height]
        x, y, w, h = ann['bbox']
        
        # Convert to integers for drawing
        start_point = (int(x), int(y))
        end_point = (int(x + w), int(y + h))
        
        # Draw the rectangle (Green, thickness of 3)
        cv2.rectangle(image, start_point, end_point, (0, 255, 0), 3)
        
        # Draw the category ID text
        cv2.putText(image, f"ID: {ann['category_id']}", (int(x), int(y) - 10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0), 2)
        found_boxes += 1

print(f"✅ Found {found_boxes} boxes for {target_image}")

# 5. Display the result
cv2.imshow('Label Check', image)
print("Press any key on the image window to close it.")
cv2.waitKey(0)
cv2.destroyAllWindows()

# Optional: Save it so you can view it outside of Python
# cv2.imwrite('debug_visual_check.png', image)
```
